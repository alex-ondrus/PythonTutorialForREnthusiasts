## Load Libraries & Data

```python
from plotnine.data import mpg
```

## Plotnine

Looking for a scatter plot of `cty` vs `hwy`, coloured by `cyl`.

```python
from plotnine import ggplot, aes, geom_point

p9_solution = (
    ggplot(mpg, aes(x = 'cty', y = 'hwy', color = 'cyl')) 
        + geom_point() 
)

p9_solution
```

## Seaborn

Looking for a violin plot showing the distribution of `hwy` for each `class`. I still do not know how to store the plotting object and display using commands the same way I can for Plotnine.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

(
    sns.violinplot(
        data=mpg,
        x="class",
        y="hwy"
    )
)

plt.title("Distribution of Highway Milage for Different Classes of Car")
plt.show()
```

## Matplotlib Solution

Using Copilot to figure out how to create a Matplotlib object with two sub-plots, it turns out that I create them as a vector in the initial construction. Controling the figure size can be done analogously to the tutorial, but only one figure size can be applied to both figures using this method. Other functions must be used to create variable sized figures.

```python
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1,2, figsize=(8,4))

ax1.hist(mpg["hwy"])
ax1.set_title("Highway Mileage")
ax1.set_xlabel("mpg")
ax2.boxplot(mpg["cty"])
ax2.set_title("City Mileage")
ax2.set_xlabel("mpg")

plt.show()
```
