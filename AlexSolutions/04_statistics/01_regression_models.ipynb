# Statistical Modeling Solutions

## Load Data

```python
import statsmodels.formula.api as smf
import pandas as pd
import seaborn as sns

df = sns.load_dataset("diamonds")
```

## Build Model and Summarize

```python
model = smf.ols(
    formula="price ~ carat + depth + table",
    data=df
).fit()

print(model.summary())
```

All predictors (`carat`, `depth`, and `table`) appear significant at the 0.05 level.

## Using `Scikit-learn`

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import train_test_split

X = df[['carat', 'depth', 'table']].dropna()
y = df.loc[X.index, 'price']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=305
)

reg = LinearRegression()
reg.fit(X_train, y_train)

y_pred = reg.predict(X_test)

print(
    f"Root Mean Squared Error: {root_mean_squared_error(y_test, y_pred):.2f}"
)
print(
    f"R2 Score: {r2_score(y_test, y_pred):.2f}"
)
```

```python
model_w_int = smf.ols(
    formula="price ~ carat + depth + table + carat * depth",
    data=df
).fit()

print(model_w_int.summary())
```
